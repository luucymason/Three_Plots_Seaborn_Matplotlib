```python
!pip install wbdata
import matplotlib.pyplot as plt
import wbdata
import pandas as pd
import numpy as np
import seaborn as sns
import seaborn.objects as so
# the keys are World Bank indicator codes, the values are the column names we want them to have
indicators = {
    "NY.GDP.PCAP.KD": "gdp_per_capita",
    "NY.GDP.PCAP.KD.ZG": "gdp_growth",
    "SP.DYN.LE00.IN": "life_expectancy",
    "SP.POP.TOTL": "population",
}
countries = ["GBR", "FRA", "ESP", "ITA", "NLD",
             "CHN", "ARE", "IND", "JOR", "USA"]
# retriving the desired data
df = wbdata.get_dataframe(indicators,
                          country=countries,
                          date=("2000", "2025"))
df = df.reset_index()
df = df.rename(columns={"date": "year"})
df["year"] = df["year"].astype(int)
df.head()
df.info() #country, year, gdp_per_capita, gdp_growth, life_expectancy, population
#my question = tracking life expectancy over time
#making 3 charts that together answer the question
#life exp of counted countries over years, colour/marker for each country
#life exp of counted countries in a specific year
#life exp of specific country vs year in sns.lineplot
#life exp in specific countries over years
(
    so.Plot(df, x="year", y="life_expectancy", color='country')
    .add(so.Line(linewidth=2), so.Agg())
    .label(x = 'Year',
           y='Mean life expectancy',
           color='country',
           title='Mean life expectancy by year')
)
from google.colab import drive
drive.mount("/content/drive")
plt.savefig("/content/drive/MyDrive/life_exp_over_years.png",
            dpi=300, bbox_inches="tight")
#life exp of countries in a specific year, bar chart used
#make column filtering by only one year
fig, ax = plt.subplots(figsize=(13, 5))
sns.barplot(data=df.query("year == 2020"), x="country", y="life_expectancy", ax=ax)
ax.set_title("Life expectancy in each country in 2020")
ax.set_xlabel("Countries measured")
ax.set_ylabel("Life expectancy")
fig.tight_layout()
plt.show()
plt.savefig("/content/drive/MyDrive/life_exp_2020.png",
            dpi=300, bbox_inches="tight")
#life exp of specific country vs year in sns.lineplot
#point at the start of covid, what could have made life expectancy dip
# Look up life expectancy in France in 2020
#have to define france in the dataset before doing this
france = df.query("country == 'France'")
y2020 = france.loc[france["year"] == 2020, "life_expectancy"].iloc[0]
fig, ax = plt.subplots(figsize=(13, 5))
sns.lineplot(data=df.query("country == 'France'"),
    x="year",
    y="life_expectancy", marker="o", ax=ax)
ax.annotate("Start of Covid-19",
    xy=(2020, y2020),
    xytext=(2016, y2020 + 1.2),
    arrowprops=dict(arrowstyle="->", color="black"),
    fontsize=10)
ax.set_title("Life expectancy in France across the years")
ax.set_xlabel("Years")
ax.set_ylabel("Life expectancy")
fig.tight_layout()
plt.show()
plt.savefig("/content/drive/MyDrive/life_exp_France.png",
            dpi=300, bbox_inches="tight")

```
